## Praktikan 1 : K-Nearest Neighbors

```python
import pandas as pd # 202431163_Maftuh Faisal Jauhari
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataseed = pd.read_csv("seed.xls",sep=";") # 202431163_Maftuh Faisal Jauhari
dataseed
```

```python
dataseed.columns # 202431163_Maftuh Faisal Jauhari
```

```python
x = dataseed.drop('class',axis = 1) # 202431163_Maftuh Faisal Jauhari
y = dataseed['class']
```

```python
x # 202431163_Maftuh Faisal Jauhari
```

```python
y # 202431163_Maftuh Faisal Jauhari
```

```python
from sklearn.model_selection import train_test_split # 202431163_Maftuh Faisal Jauhari

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 0)
```

```python
from sklearn.neighbors import KNeighborsClassifier # 202431163_Maftuh Faisal Jauhari
```

```python
knn = KNeighborsClassifier(n_neighbors=7) # 202431163_Maftuh Faisal Jauhari
knn.fit(x_train, y_train)
```

```python
y_pred = knn.predict(x_test) # 202431163_Maftuh Faisal Jauhari
```

```python
x_new = np.array([[15,25.5,7,30,11,5,10]]) # 202431163_Maftuh Faisal Jauhari
prediksi = knn.predict(x_new)
print(prediksi)
```

```python
from sklearn.metrics import accuracy_score # 202431163_Maftuh Faisal Jauhari

akurasi = accuracy_score(y_test, y_pred)
print(akurasi)
```

```python
k_values = range(1,20) # 202431163_Maftuh Faisal Jauhari
accuracies = []

for k in k_values :
    model = KNeighborsClassifier(n_neighbors = k)
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    acc = accuracy_score(y_test, y_pred)
    accuracies.append(acc)

plt.figure(figsize = (5,7))
plt.plot(k_values, accuracies)
plt.xlabel("Nilai K")
plt.ylabel("nilai_akurasi")
plt.title("Pemilihan K pada KNN")
plt.grid(True)
plt.show()
```

## Praktikan 2: Support Vector Machine

```python
import pandas as pd # 202431163_Maftuh Faisal Jauhari
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
dataset = pd.read_csv('iris-flower.csv') # 202431163_Maftuh Faisal Jauhari
dataset
```

```python
dataset['Species'].unique() # 202431163_Maftuh Faisal Jauhari
```

```python
le = LabelEncoder() # 202431163_Maftuh Faisal Jauhari
dataset['Species_encoded'] = le.fit_transform(dataset['Species'])

x = dataset.drop(columns=['Id', 'Species', 'Species_encoded'], errors = 'ignore')
y = dataset['Species_encoded']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3, random_state=0) # 202431163_Maftuh Faisal Jauhari
```

```python
pipeline = Pipeline([('scaler', StandardScaler()),('svc',SVC())]) # 202431163_Maftuh Faisal Jauhari
```

```python
params = { # 202431163_Maftuh Faisal Jauhari
    'SVC__kernel': ['poly', 'rbf', 'sigmoid'],
    'SVC__C': [0.5, 1, 10, 100],
    'SVC__gamma': ['scale', 1, 0.1, 0.001]
}
```

```python
grid_search = GridSearchCV( # 202431163_Maftuh Faisal Jauhari
    estimator=pipeline,
    param_grid=params,
    scoring='accuracy',
    cv=5,
    n_jobs=1,
    verbose=1
)
```

```python
from sklearn.model_selection import GridSearchCV # 202431163_Maftuh Faisal Jauhari
from sklearn.metrics import accuracy_score

params = {
    'svc__kernel': ['poly', 'rbf', 'sigmoid'],
    'svc__C': [0.5, 1, 10, 100],
    'svc__gamma': ['scale', 1, 0.1, 0.001]
}

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=params,
    scoring='accuracy',
    cv=5,
    n_jobs=1,
    verbose=1
)

grid_search.fit(x_train, y_train)

best_model = grid_search.best_estimator_

y_train_pred = best_model.predict(x_train)
y_test_pred = best_model.predict(x_test)

train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_test_pred)

print("Train Accuracy:", train_accuracy)
print("Test Accuracy:", test_accuracy)
```

```python
print("\nClassification Report (Test):\n", # 202431163_Maftuh Faisal Jauhari
    classification_report(y_test, y_test_pred, target_names=le.classes_))

print(
    "\nConfusion Matrix (Test):\n", confusion_matrix(y_test, y_test_pred))
```

```python
iris_anonim = np.array([[6.5, 3.0, 7.0, 4.8]]) # 202431163_Maftuh Faisal Jauhari
pred = best_model.predict(iris_anonim)
print("Predicted label (encoded): ", pred)
print("Predicted Species : ",le.inverse_transform(pred))
```
