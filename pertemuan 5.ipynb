# Praktikum 1: Regresi

## Import Library

```python
import numpy as np # 202431163_Maftuh Faisal Jauhari
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
df = pd.DataFrame([[153,63], [174,81], [136,56], # 202431163_Maftuh Faisal Jauhari
                   [186,91], [128,47], [136,57],
                   [179,76], [163,72], [131,48]])
df.columns = ["x", "y"]
```

```python
df.corr() # 202431163_Maftuh Faisal Jauhari
```

## Membagi Data

```python
# 202431163_Maftuh Faisal Jauhari
x_train = df['x'].values[:,np.newaxis]
y_train = df['y'].values

x_train
```

```python
y_train # 202431163_Maftuh Faisal Jauhari
```

## Model Training

```python
# 202431163_Maftuh Faisal Jauhari
lm = LinearRegression()
lm.fit(x_train, y_train)
```

## Membuat Prediksi

```python
# 202431163_Maftuh Faisal Jauhari
x_test = [[170], [160], [180], [171], [150]]
p = lm.predict(x_test)
p
```

## Koefisien dan Intersep

```python
# 202431163_Maftuh Faisal Jauhari
print("coefficient :", lm.coef_)
print("intercept :", lm.intercept_)
```

## Visualisasi Model

```python
# 202431163_Maftuh Faisal Jauhari
pb = lm.predict(x_train)
dfc = pd.DataFrame({"x" : df["x"], "y" : pb})
plt.scatter(df['x'], df['y'])
plt.plot(dfc['x'], dfc['y'], color="magenta", linewidth=1)
plt.xlabel('Nilai X')
plt.ylabel('Nilai Y')
```

## Implementasi Pada Dataset

```python
df = pd.read_csv('botak.csv')
df # 202431163_Maftuh Faisal Jauhari
```

```python
df.info() # 202431163_Maftuh Faisal Jauhari
```

```python
# 202431163_Maftuh Faisal Jauhari
num_df = df.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(10,6))
sns.heatmap(num_df.corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
db_clean = df.dropna(subset=["umur", "gaji", "tinggi", "berat", "botak_prob"])

x = db_clean[["umur", "gaji", "tinggi", "berat"]]
y = db_clean["botak_prob"]
# 202431163_Maftuh Faisal Jauhari
```

```python
# 202431163_Maftuh Faisal Jauhari
from sklearn.model_selection import train_test_split as tts

x_train, x_test, y_train, y_test = tts(x, y, test_size=0.2, random_state=0)
```

```python
model = LinearRegression()
# 202431163_Maftuh Faisal Jauhari
model.fit(x_train, y_train)
```

## Prediksi

```python
# 202431163_Maftuh Faisal Jauhari
y_pred = model.predict(x_test)
```

## Evaluasi

```python
# 202431163_Maftuh Faisal Jauhari
from sklearn.metrics import mean_squared_error, r2_score

print("coefficient :", model.coef_)
print("intercept :", model.intercept_)
print("MSE :", mean_squared_error(y_test, y_pred))
print("R2 Score :", r2_score(y_test, y_pred))
```

## Visualisasi Model

```python
# 202431163_Maftuh Faisal Jauhari
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha = 0.5)
plt.xlabel("Nilai asli y_test")
plt.ylabel("Nilai prediksi y_test")
plt.title("Nilai asli Vs Nilai prediksi")
plt.plot([0,1],[0,1],'r--')
plt.show()
```

# Praktikum 2: Decision Tree

## Import Library

```python
import numpy as np # 202431163_Maftuh Faisal Jauhari
import pandas as pd
import matplotlib.pyplot as plt
```

## Pre Processing

```python
df = pd.read_csv('iris-flower.csv', sep=",")
df # 202431163_Maftuh Faisal Jauhari
```

```python
df.info() # 202431163_Maftuh Faisal Jauhari
```

```python
df['Id'].unique() # 202431163_Maftuh Faisal Jauhari
```

```python
df['Species'].unique() # 202431163_Maftuh Faisal Jauhari
```

```python
from sklearn.preprocessing import LabelEncoder
# 202431163_Maftuh Faisal Jauhari
le = LabelEncoder()
df['Species'] = le.fit_transform(df['Species'])
print(df['Species'].unique())

for i, label in enumerate(le.classes_):
    print(f"(i) = {label}")
```

```python
df # 202431163_Maftuh Faisal Jauhari
```

## Modelling

```python
x = df.drop(['Id', 'Species'], axis=1)
x # 202431163_Maftuh Faisal Jauhari
```

```python
y = df['Species']
y # 202431163_Maftuh Faisal Jauhari
```

```python
from sklearn.model_selection import train_test_split # 202431163_Maftuh Faisal Jauhari

x_train, x_test, y_train, y_test, = train_test_split(x, y, test_size = 0.25, random_state=0)
```

```python
from sklearn.tree import DecisionTreeClassifier # 202431163_Maftuh Faisal Jauhari

classifier = DecisionTreeClassifier()
classifier.fit(x_train, y_train)
```

```python
from sklearn import tree # 202431163_Maftuh Faisal Jauhari

plt.subplots(figsize = (15,10))
tree.plot_tree(classifier, fontsize=10) 
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier # 202431163_Maftuh Faisal Jauhari

classifier = DecisionTreeClassifier(criterion="entropy", max_depth=5, random_state=42)
classifier.fit(x_train, y_train)
```

```python
from sklearn import tree # 202431163_Maftuh Faisal Jauhari

plt.subplots(figsize = (15,10))
tree.plot_tree(classifier, fontsize=10) 
plt.show()
```

```python
depth = classifier.get_depth()
depth # 202431163_Maftuh Faisal Jauhari
```

## Evaluasi

```python
from sklearn.metrics import accuracy_score # 202431163_Maftuh Faisal Jauhari

y_pred = classifier.predict(x_test)
akurasi = accuracy_score(y_test, y_pred)
print(akurasi)
```
