## Adjustment of Variable Types

```python
df.dtypes

# 202431163_Maftuh Faisal Jauhari
```

```python
cat_col = ["gender", "ever_married", "work_type", "Residence_type", "smoking_status"]

for col in cat_col:
    df[col] = df[col].astype('category')

df.dtypes

# 202431163_Maftuh Faisal Jauhari
```

## Detection and Treatment of Missing Values

```python
df.isnull().sum()

# 202431163_Maftuh Faisal Jauhari
```

```python
df['bmi'] = df['bmi'].fillna(df['bmi'].median())

df.head()

# 202431163_Maftuh Faisal Jauhari
```

```python
df.isnull().sum()

# 202431163_Maftuh Faisal Jauhari
```

```python
sns.boxplot(x="bmi", data=df)
plt.title("Distribusi Kelas BMI")
plt.show()

# 202431163_Maftuh Faisal Jauhari
```

```python
Q1 = df['age'].quantile(1/4)
Q3 = df['age'].quantile(3/4)

IQR = Q3-Q1

#boundaries
lower_bound = Q1 - 3/2 * IQR
upper_bound = Q3 + 3/2 * IQR

outliers = df[(df["age"] < lower_bound) | (df["age"] > upper_bound)]
print("Jumlah outlier umur: ", len(outliers))

# 202431163_Maftuh Faisal Jauhari
```

## Correlation of Variables

```python
num_df = df.select_dtypes(include=['int64','float64'])

plt.figure(figsize=(10,6))
sns.heatmap(num_df.corr(), annot=True, cmap="coolwarm") # corr() returns
plt.show

# 202431163_Maftuh Faisal Jauhari
```
